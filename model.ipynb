# Data Cleaning & Preprocessing

```python
df.isnull().sum()
```

```python
df['Perceived_Stress_Score']=df['Perceived_Stress_Score'].fillna(
                                df['Perceived_Stress_Score'].mean())
```

```python
df.isnull().sum()
```

```python
df['Academic_Performance_GPA']=df['Academic_Performance_GPA'].fillna(
                                df['Academic_Performance_GPA'].mean())
```

```python
df.isnull().sum()
```

```python
df_cleaned=df.copy()
```

```python
df_cleaned.describe()
```

```python
num_cols = df.select_dtypes(include='number').columns
cat_cols = df.select_dtypes(include=['object', 'category', 'bool']).columns

print("Numerical columns:", len(num_cols))
print("Categorical columns:", len(cat_cols))

print("\nNumerical:", list(num_cols))
print("\nCategorical:", list(cat_cols))
```

```python
for col in cat_cols:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))
```

```python
data_preprocessed = df_cleaned.copy()
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

cols_to_encode = ['Primary_Platform', 'Device_Type']

encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

encoded_data = encoder.fit_transform(data_preprocessed[cols_to_encode])

encoded_df = pd.DataFrame(
    encoded_data,
    columns=encoder.get_feature_names_out(cols_to_encode),
    index=data_preprocessed.index
)

data_preprocessed = pd.concat(
    [data_preprocessed.drop(columns=cols_to_encode), encoded_df],
    axis=1
)

print(data_preprocessed.head())
```

```python
from sklearn.preprocessing import StandardScaler

# Select numerical columns
num_cols = data_preprocessed.select_dtypes(include='number').columns.tolist()

# Exclude target and ID columns
exclude_cols = ['Student_ID', 'Overall_Impact']
num_cols = [col for col in num_cols if col not in exclude_cols]

# Standardization
scaler = StandardScaler()

data_preprocessed[num_cols] = scaler.fit_transform(
    data_preprocessed[num_cols]
)
```

```python
data_preprocessed.head()
```

```python
data_preprocessed = data_preprocessed.drop(columns=['Student_ID'])
```

```python
# Convert Gender to Boolean
if 'Gender' in data_preprocessed.columns:
    data_preprocessed['IsMale'] = (
        data_preprocessed['Gender']
        .astype(str).str.strip().str.lower() == 'male'
    )
    data_preprocessed.drop(columns=['Gender'], inplace=True)

# One-hot encode remaining categorical columns
cols = [
    col for col in [
        'Academic_Level',
        'Social_Comparison_Frequency'
    ]
    if col in data_preprocessed.columns
]

if cols:
    data_preprocessed = pd.get_dummies(
        data_preprocessed,
        columns=cols,
        dtype=int
    )
```

# Test & Train Data Set Spliting

```python
Y=data_preprocessed['Overall_Impact']
```

```python
X = data_preprocessed.drop(
    columns=['Overall_Impact', 'Student_ID'],
    errors='ignore'
)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.20, random_state=42
)
```

# Training the Model

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score
)
```

```python
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight='balanced'
)
```

```python
print(X_train.dtypes)
print("Missing values:", X_train.isnull().sum().sum())
print("Target missing values:", Y_train.isnull().sum())
```

```python
model.fit(X_train, Y_train)
```

# Predicting Output

```python
predictions=model.predict(X_test)
```

```python
count=0
for i in range(len(predictions)):
    if predictions[i]==Y_test.iloc[i]:
        count=count+1
    else:
        print(f"Prediction : {predictions[i]} |||| Actual : {Y_test.iloc[i]}")
print(count/len(predictions)*100,"%")
```

```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
    log_loss
)

# 1. Predictions and probabilities
predictions = model.predict(X_test)
probabilities = model.predict_proba(X_test)

# 2. Overall performance
print("=" * 50)
print("       RANDOM FOREST MODEL EVALUATION")
print("=" * 50)

print(f"Accuracy:          {accuracy_score(Y_test, predictions):.4f}")
print(f"Balanced Accuracy: {balanced_accuracy_score(Y_test, predictions):.4f}")
print(f"Precision (Macro): {precision_score(Y_test, predictions, average='macro', zero_division=0):.4f}")
print(f"Recall (Macro):    {recall_score(Y_test, predictions, average='macro', zero_division=0):.4f}")
print(f"F1 Score (Macro):  {f1_score(Y_test, predictions, average='macro', zero_division=0):.4f}")
print(f"Log Loss:          {log_loss(Y_test, probabilities, labels=model.classes_):.4f}")

# 3. Detailed classification report
print("\nCLASSIFICATION REPORT")
print(classification_report(Y_test, predictions, zero_division=0))

# 4. Confusion matrix
cm = confusion_matrix(Y_test, predictions, labels=model.classes_)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=model.classes_,
    yticklabels=model.classes_
)
plt.title("Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.tight_layout()
plt.show()

# 5. Correct vs incorrect predictions
correct = np.asarray(Y_test) == predictions

print("\nPREDICTION SUMMARY")
print("Correct predictions:", correct.sum())
print("Incorrect predictions:", (~correct).sum())
print(f"Error rate: {np.mean(~correct) * 100:.2f}%")

# 6. Show misclassified examples
errors = pd.DataFrame({
    "Actual": np.asarray(Y_test),
    "Predicted": predictions
}, index=Y_test.index)

errors = errors[errors["Actual"] != errors["Predicted"]]

print("\nMISCLASSIFIED EXAMPLES (first 10)")
display(errors.head(10))

# 7. Feature importance
importance = pd.Series(
    model.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

print("\nTOP 10 IMPORTANT FEATURES")
display(importance.head(10).to_frame("Importance"))

plt.figure(figsize=(8, 5))
importance.head(10).sort_values().plot(kind='barh')
plt.title("Top 10 Feature Importances")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()
```
